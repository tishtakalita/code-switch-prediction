```python
import json
import numpy as np
import matplotlib.pyplot as plt
```

```python
delta_sec = [0.05, 0.10, 0.15, 0.20, 0.25, 0.30, 0.40, 0.50, 0.75, 1.00, 1.50, 2.00, 3.00, 4.00, 5.00]
```

```python
with open("outputs/fs2_pre_switch_validation.json", "r") as f:
    results = json.load(f)
```

```python
mean_roc_auc = []
mean_ap = []
mean_log_loss = []
mean_brier = []

for delta in results:

    metrics = [
        results[delta][layer]["metrics"]
        for layer in results[delta]
    ]

    mean_roc_auc.append(np.mean([metric["roc_auc"] for metric in metrics]))
    mean_ap.append(np.mean([metric["ap"] for metric in metrics]))
    mean_log_loss.append(np.mean([metric["log_loss"] for metric in metrics]))
    mean_brier.append(np.mean([metric["brier_score"] for metric in metrics]))
```

```python
plt.figure(figsize=(20, 5))
plt.plot(delta_sec, mean_roc_auc, marker="o")

plt.xlabel("Prediction horizon (sec)")
plt.ylabel("Mean ROC-AUC")

plt.xticks(delta_sec, rotation=50)

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20, 5))
plt.plot(delta_sec, mean_ap, marker="o")

plt.xlabel("Prediction horizon (sec)")
plt.ylabel("Mean Average Precision")

plt.xticks(delta_sec, rotation=50)

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20, 5))
plt.plot(delta_sec, mean_log_loss, marker="o")

plt.xlabel("Prediction horizon (sec)")
plt.ylabel("Mean Log Loss")

plt.xticks(delta_sec, rotation=50)

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20, 5))
plt.plot(delta_sec, mean_brier, marker="o")

plt.xlabel("Prediction horizon (sec)")
plt.ylabel("Mean Brier Score")

plt.xticks(delta_sec, rotation=50)

plt.grid(True)
plt.tight_layout()
plt.show()
```
